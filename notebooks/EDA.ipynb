```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('../vehicles.csv')

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
missing_percent = (df.isnull().mean()*100).sort_values(ascending=False)
missing_percent
```

## Análise de valores ausentes

Foi realizada uma investigação dos valores ausentes presentes no conjunto de dados.

As colunas com maior quantidade de dados faltantes foram `is_4wd`, `paint_color`, `odometer`, `cylinders` e `model_year`.

A coluna `is_4wd` apresenta aproximadamente 50% de valores ausentes, indicando que essa informação não está disponível para uma grande parte dos veículos. Também foram identificados valores ausentes em características importantes como quilometragem (`odometer`) e ano do modelo (`model_year`).

Neste momento, os valores ausentes não serão removidos, pois a exclusão desses registros poderia reduzir significativamente o tamanho do conjunto de dados. A decisão de preenchimento ou remoção dependeria do objetivo específico da análise.

```python
missing = missing_percent[missing_percent > 0].sort_values(ascending=False)

fig = px.bar(
    x=missing.index,
    y=missing.values,
    title="Percentual de Valores Ausentes por Coluna",
    labels={
        "x": "Coluna",
        "y": "Percentual (%)"
    },
    text=missing.values
)

fig.update_traces(
    texttemplate='%{text:.1f}%',
    textposition='outside'
)

fig.show()
```

## Análise da quilometragem

O histograma apresenta a distribuição da quilometragem dos veículos anunciados.

É possível observar que a maior parte dos veículos possui quilometragem concentrada abaixo de aproximadamente 200 mil km. Também existem alguns veículos com valores muito elevados de odômetro, próximos de 1 milhão de quilômetros, que podem representar casos extremos dentro do conjunto de dados.

```python
df_filtered = df[df['odometer'] < 400000]

fig = px.histogram(
    df_filtered,
    x='odometer',
    title='Distribuição da Quilometragem dos Veículos (até 400 mil km)',
    nbins=50
)

fig.show()
```

## Relação entre Quilometragem e Preço dos Veículos Usados

O gráfico de dispersão apresenta a relação entre a quilometragem dos veículos (odômetro) e seus respectivos preços.

É possível observar uma tendência de redução do preço conforme a quilometragem aumenta, indicando que veículos mais rodados geralmente possuem menor valor de mercado. Também existem alguns pontos fora do padrão, representando veículos com preços elevados mesmo apresentando alta quilometragem, que podem estar relacionados a características específicas como marca, modelo ou estado de conservação.


```python
fig = px.scatter(
    df,
    x='odometer',
    y='price',
    title='Relação entre Quilometragem e Preço dos Veículos Usados',
    labels={
        'odometer': 'Quilometragem (km)',
        'price': 'Preço (US$)'
    },
    opacity=0.5,
)

fig.show()
```

## Distribuição dos Tipos de Veículos

Este gráfico apresenta a quantidade de anúncios para cada categoria de veículo presente no conjunto de dados.

É possível observar quais tipos de veículos possuem maior quantidade de registros e quais aparecem com menor frequência nos anúncios. Essa análise ajuda a entender o perfil dos veículos disponíveis na base e identificar as categorias mais comuns no mercado analisado.

```python
type_count = df['type'].value_counts().reset_index()
type_count.columns = ['type', 'quantidade']

fig = px.bar(
    type_count,
    x='type',
    y='quantidade',
    title='Quantidade de Anúncios por Tipo de Veículo',
    labels={
        'type': 'Tipo de Veículo',
        'quantidade': 'Quantidade de Anúncios'
    }
)

fig.show()
```

## Análise da distribuição de preços

O histograma apresenta a distribuição dos preços dos veículos anunciados.

É possível observar que a maioria dos veículos possui preços concentrados em valores mais baixos, enquanto poucos anúncios apresentam valores muito elevados. Esses valores extremos influenciam a distribuição e podem representar veículos com características diferenciadas, como modelos mais novos ou de maior valor de mercado.

```python
fig = px.histogram(
    df,
    x='price',
    title='Distribuição de Preços dos Veículos',
    nbins=50
)

fig.show()
```

## Preço Médio por Condição do Veículo

Este gráfico apresenta o preço médio dos veículos de acordo com sua condição de conservação.

A análise permite comparar como o estado do veículo está relacionado ao seu valor médio anunciado. É possível observar diferenças entre as categorias de condição, indicando que veículos classificados em melhores condições tendem a apresentar valores médios diferentes em relação às demais categorias.

Essa comparação ajuda a entender se a condição informada no anúncio possui relação com o preço de mercado dos veículos.

```python
condition_price = (
    df.groupby("condition", as_index=False)["price"].mean()
)

fig = px.bar(
    condition_price,
    x='condition',
    y='price',
    title='Preço Médio por Condição do Veículo',
    labels={
        'condition': 'Condição do Veículo',
        'price': 'Preço Médio (US$)'
    },
    text='price'
)

fig.update_traces(
    texttemplate='%{text:.0f}',
    textposition='outside'
)

fig.show()
```
